## 4. Data Cleaning & Preparation :
                    In this step, we cleaned the raw dataset to make it ready for analysis by handling missing values, removing unwanted columns, and treating outliers.

#### 4.1 Dropping Unnecessary Columns
Removed Unnamed: 0 (an auto-generated index column) and car_name (not useful for numerical analysis) to keep only relevant features.

#### 4.2 Handling Missing Values
- Text Columns — Filled with 'Unknown'

brand and model columns had missing values that cannot be logically guessed, so they were filled with 'Unknown' to preserve the rows.

- Categorical Columns — Filled with Mode

seller_type, fuel_type, transmission_type — filled with the most frequently occurring value, which is the standard approach for categorical data.

-  Numerical Columns — Filled with Median

vehicle_age, km_driven, mileage, engine, max_power — filled with the median instead of mean, because median is resistant to outliers.

-  Seats Column — Filled with Mode

seats is a discrete feature, so it was filled with the most common seat count.

-  Target Column — selling_price :  Since selling_price is the target variable, it cannot be imputed.
Checked the percentage of missing values — it was less than 6% (well under the 20% threshold).
Dropped those rows safely without losing significant data.


#### 4.3 Feature Engineering

Created a new column price_lakh by converting selling_price to Lakhs (÷ 1,00,000) for easier interpretation and visualization.


#### 4.4 Removing Duplicates

Checked for duplicate rows using .duplicated().sum()
Dropped all duplicate entries using .drop_duplicates() to ensure data integrity.



```python
# Drop unnecessary columns: 'Unnamed: 0' (index-like) and 'car_name' (not useful for analysis)
df.drop(columns=['Unnamed: 0','car_name'],inplace = True)
```

```python
# Fill missing values in 'brand' column with 'Unknown' since we cannot assign other brand names
df['brand'].fillna('Unknown', inplace=True)
```

```python
# Fill missing values in 'model' column with 'Unknown'
df['model'].fillna('Unknown', inplace=True)
```

```python
# Check for missing values again after filling
df.isna().sum()
```

## For Categorical Columns :

```python
# Fill missing values in categorical columns with the mode (most frequent value)
cat_cols = ['seller_type', 'fuel_type', 'transmission_type']

for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

## For Numerical Columns :

```python
# Fill missing values in numerical columns with the median
num_cols = ['vehicle_age','km_driven','mileage','engine','max_power']

for col in num_cols:
    df[col].fillna(df[col].median(), inplace=True)
```

```python
# Fill missing values in 'seats' column with the mode (most frequent value)
df['seats'].fillna(df['seats'].mode()[0], inplace=True)
```

## selling_price : TARGET COLUMN

- This column cannot be modified as it is a target column.
- We will check first how much data is missing from the column.
- If the data missing is less than 20 % we can drop the rows containing null values.

```python
# Check the number of missing values in the target column 'selling_price'
df['selling_price'].isna().sum()
```

```python
# Drop rows with missing values in 'selling_price' since it's the target and missing data is minimal (about 5-6%)
df = df.dropna(subset=['selling_price'])
```

## Dealing with the Outliers

- There are outliers present in the 'selling_price' column. But it is the "Target Column". Therefore,Outliers in selling price were retained because     they represent real-world variations such as luxury vehicles.
- And for the 'engine' & 'max_power' column , High values can exist depending on the size of the vehicles.

```python
# Handle outliers in 'km_driven' column using the IQR method to cap extreme values
Q1 = df['km_driven'].quantile(0.25)
Q3 = df['km_driven'].quantile(0.75)
IQR = Q3 - Q1

df['km_driven'] = df['km_driven'].clip(Q1 - 1.5*IQR, Q3 + 1.5*IQR)
```

# 4. 🟢 Price Segmentation Analysis

Categorizing cars into price segments to understand market distribution.

SegmentPrice :  Range
💚 BudgetBelow ₹10 Lakh
💛 Mid-range₹10 – ₹20 Lakh
🧡 Premium₹20 – ₹40 Lakh
❤️ LuxuryAbove ₹40 Lakh

Created a new column price_segment based on the above categories.
Visualized using both a bar chart (count per segment) and a pie chart (market share %).


💡 Insight: The Budget segment dominates the market with the highest number of cars. The Luxury segment is a small but distinct niche, representing premium vehicles.

```python
## Price segmentation analysis
### Price Segmentation: Budget, Mid-range, Premium, Luxury

# Create price categories
def categorize_price(price):
    if price < 10:
        return 'Budget'
    elif price < 20:
        return 'Mid-range'
    elif price < 40:
        return 'Premium'
    else:
        return 'Luxury'

df['price_segment'] = df['price_lakh'].apply(categorize_price)

# Count cars in each segment
segment_counts = df['price_segment'].value_counts().reindex(['Budget', 'Mid-range', 'Premium', 'Luxury']).reset_index()
segment_counts.columns = ['Segment', 'Count']

seg_colors = ['#6C5CE7', '#00CEC9', '#FDCB6E', '#FD79A8']

# Plotly subplots: Bar + Pie
from plotly.subplots import make_subplots
fig = make_subplots(rows=1, cols=2, specs=[[{'type': 'bar'}, {'type': 'pie'}]],
                    subplot_titles=['<b>Cars by Price Segment</b>', '<b>Market Share by Segment</b>'])

fig.add_trace(go.Bar(
    x=segment_counts['Segment'], y=segment_counts['Count'],
    marker_color=seg_colors, text=segment_counts['Count'],
    textposition='outside', textfont_size=13,
    showlegend=False
), row=1, col=1)

fig.add_trace(go.Pie(
    labels=segment_counts['Segment'], values=segment_counts['Count'],
    marker=dict(colors=seg_colors, line=dict(color='#1e1e2e', width=2)),
    textinfo='percent+label', textfont_size=12,
    pull=[0.04]*4, hole=0.35
), row=1, col=2)

fig.update_layout(
    title='<b>Price Segmentation Analysis</b>',
    title_font_size=20, title_x=0.5,
    width=1100, height=500,
    font=dict(family='Inter, sans-serif'),
    plot_bgcolor='rgba(0,0,0,0)',
    showlegend=False,
)
fig.show()
```

### Insight: Budget segment dominates with majority of cars. Luxury segment represents only premium vehicles.

# 6. 🟣 Missing Data Visualization (Data Quality Assessment)

A final check to validate the quality of our cleaned dataset.


Computed data completeness percentage for every column.
Color-coded bars:

🟢 Green — 99%+ completeness
🔵 Blue — 95–99% completeness
🔴 Red — below 95% completeness


Also plotted a summary bar chart showing total rows, complete rows, rows with missing values, and total columns.


✅ Insight: The dataset achieved 100% completeness across all columns after cleaning — confirming that our data preparation was thorough and the dataset is fully ready for modeling.

```python
## Missing Data Visualization
### Data Completeness: Missing Values Analysis

completeness = (1 - df.isna().sum() / len(df)) * 100
completeness = completeness.sort_values(ascending=True).reset_index()
completeness.columns = ['Column', 'Completeness']

bar_colors = ['#00B894' if x >= 99 else '#74B9FF' if x >= 95 else '#D63031' for x in completeness['Completeness']]

summary_data = {
    'Total Rows': len(df),
    'Complete Cols': int((df.isna().sum() == 0).sum()),
    'Cols w/ Missing': int((df.isna().sum() > 0).sum()),
    'Total Columns': len(df.columns)
}

from plotly.subplots import make_subplots
fig = make_subplots(rows=1, cols=2,
                    subplot_titles=['<b>Data Completeness by Column</b>', '<b>Dataset Summary</b>'])

fig.add_trace(go.Bar(
    y=completeness['Column'], x=completeness['Completeness'].round(1),
    orientation='h', marker_color=bar_colors,
    text=completeness['Completeness'].apply(lambda v: f'{v:.1f}%'),
    textposition='outside', textfont_size=11,
    showlegend=False
), row=1, col=1)

fig.add_trace(go.Bar(
    x=list(summary_data.keys()), y=list(summary_data.values()),
    marker_color=['#00CEC9', '#00B894', '#D63031', '#74B9FF'],
    text=[str(v) for v in summary_data.values()],
    textposition='outside', textfont_size=12,
    showlegend=False
), row=1, col=2)

fig.update_xaxes(range=[90, 101], row=1, col=1)
fig.update_layout(
    title='<b>Data Quality Overview</b>',
    title_font_size=20, title_x=0.5,
    width=1150, height=550,
    font=dict(family='Inter, sans-serif'),
    plot_bgcolor='rgba(0,0,0,0)',
)
fig.show()
```

### Insight: Dataset has 100.0% minimum completeness and 100.0% maximum completeness.
### Total 0 rows contain missing values that were handled during cleaning.
